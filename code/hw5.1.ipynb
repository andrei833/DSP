```python
import numpy as np
import matplotlib.pyplot as plt
dt = 0.01
t = np.arange(0,2,0.01)
x = np.cos(np.pi * 2 * 2 * t) + np.cos (2 * np.pi * 20 * t)

X = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x)))*dt
f = np.fft.fftshift(np.fft.fftfreq(len(t),dt))

X_filtered = X.copy()
X_filtered[np.abs(f) > 10] = 0

x_filtered = np.real(np.fft.ifft(np.fft.ifftshift(X_filtered))) / dt

x_filtered = np.real(np.fft.ifft(np.fft.ifftshift(X_filtered))) / dt

fig, (ax1, ax2) = plt.subplots(2, 1)
ax1.plot(t, x, label='Original')
ax1.plot(t, x_filtered, label='Filtered')
ax1.legend()

ax2.plot(f, np.abs(X), label='Original spectrum')
ax2.plot(f, np.abs(X_filtered), label='Filtered spectrum')
ax2.legend()
plt.show()
```

```python
df = 0.1
f = np.arange(-20,20,df)
f0 = 5
X = np.where(abs(f) < f0, f0**2-f**2,0) #generate parabolic

x = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(X))) / df # compute ifft
t = np.fft.fftshift(np.fft.fftfreq(len(f), df))             #steps for the ifft


fs = 15 
T = 1/fs
t_sampled = np.arange(-5,5,T) #sample points
x_sampled = np.interp(t_sampled, t, np.real(x)) #sample values

X_sampled = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x_sampled))) * T # freqency domain
f_sampled = np.fft.fftshift(np.fft.fftfreq(len(x_sampled), T))           # steps for frequency

X_reconstructed = X_sampled.copy()
X_reconstructed[np.abs(f_sampled) > fs/2] = 0 # go trough sampled points and cut off anythings that is at f*2 > fs

x_reconstructed = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(X_reconstructed))) / T   #convert to time domain

plt.plot(t, np.real(x), label='Original')
plt.plot(t_sampled, np.real(x_reconstructed), label='Reconstructed')
plt.xlim(-2, 2)
plt.legend()
plt.show()
```
