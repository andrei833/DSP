```python
import numpy as np
import matplotlib.pyplot as plt

# step for the t line
dt = 0.001
t = np.arange(0,2,dt)

# define frequencies
f = 5
fs = 12
fa = f - fs


#generate the descrete t axis
t_sampled = np.arange(0,2,1/fs)

x = np.cos(np.pi * 2 * f * t)
x_sampled = np.cos(np.pi * 2 * f * t_sampled)
x_alias = np.cos(np.pi * 2 * fa * t)

plt.plot(t,x)
plt.plot(t_sampled,x_sampled,"ko")
plt.plot(t,x_alias)
```

```python
dt = 0.001
t = np.arange(-5,5,dt)
x = np.exp(-t**2)

X = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x))) * dt

f = np.fft.fftshift(np.fft.fftfreq(len(x),dt))
X_analytical = np.sqrt(np.pi) * np.exp(-np.pi**2 * f**2)

plt.xlim(-1,1)
plt.plot(f, np.abs(X), label="FFT")
plt.plot(f, X_analytical, "--", label="Analytical")
plt.legend()
```

```python
dt = 0.001
t = np.arange(-1,1,dt)
x = np.where(np.abs(t) <= 1/2, 1, 0)

X = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x))) * dt
f = np.fft.fftshift(np.fft.fftfreq(len(t),dt))

X_math = np.where( f == 0, 1, np.sin(np.pi * f)/(np.pi*f))

plt.plot(f, np.real(X), "--" , label='FFT')
plt.plot(f, X_math, label='Analytical')
# plt.xlim(-10, 10)
# plt.ylim(0.9, 1.1)
plt.legend()
plt.show()

```

```python
dt = 0.01
t = np.arange(-5,5,dt)
x = np.sin(np.pi*t)/(np.pi*t)

X = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x)))*dt
f = np.fft.fftshift(np.fft.fftfreq(len(t),dt))

X_math = np.where(np.abs(f) <= 0.5,1 ,0)

plt.plot(f, np.real(X), "--" , label='FFT')
plt.plot(f, X_math, label='Analytical')
plt.xlim(-2, 2)
plt.legend()
plt.show()
```

```python
dt = 0.001
t = np.arange(-5,5,dt)
x = 1/(1+t**2)

X = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(x)))*dt
f = np.fft.fftshift(np.fft.fftfreq(len(t),dt))

X_math = np.pi*np.exp(-2*np.pi*np.abs(f))

plt.plot(f, np.real(X), "--" , label='FFT')
plt.plot(f, X_math, label='Analytical')
plt.xlim(-0.4, 0.4)
plt.legend()
plt.show()
```
